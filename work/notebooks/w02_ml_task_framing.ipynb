## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

data = {
    "passage_id": [1, 2, 3, 4],
    "paper_id": ["P001", "P001", "P002", "P002"],
    "text": [
        "Previous studies achieved high accuracy, but they used only static datasets.",
        "The proposed model uses a transformer-based architecture for classification.",
        "The generalizability of current methods across different datasets remains unclear.",
        "The experiments show improved performance compared with the baseline model."
    ],
    "label": [1, 0, 1, 0]
}

df = pd.DataFrame(data)

print("One row = one research-paper passage")
display(df)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
keywords = [
    "research gap",
    "not been studied",
    "remains unclear",
    "limited evidence",
    "little is known",
    "has not been examined"
]

keyword_match = df["text"].str.lower().apply(
    lambda text: any(keyword in text for keyword in keywords)
)

print("Total passages:", len(df))
print("Passages matched by fixed keywords:", keyword_match.sum())
print("Percentage matched:", round(keyword_match.mean() * 100, 2), "%")
```
